# Pricer Finder

My app helps a small indie studio decide what price to launch its new game by comparing prices, ratings, and estimated revenues of similar Steam games in the studio’s genre.


```python
from google.colab import drive
drive.mount("/content/drive")

import glob
import os
import time
import pandas as pd

# Look for steam_games.csv in My Drive and in folders up to 3 levels down
found = []
for pattern in ["", "*/", "*/*/", "*/*/*/"]:
    found = found + glob.glob("/content/drive/MyDrive/" + pattern + "steam_games.csv")

if len(found) == 0:
    print("❌ Can't find steam_games.csv in your Drive. Upload it next to this notebook and run this cell again.")
else:
    FOLDER = os.path.dirname(found[0]) + "/"
    games = pd.read_csv(FOLDER + "steam_games.csv")
    print(f"✅ Loaded {len(games)} games from:", FOLDER)
```

```python
def median_price(games, genre):
    has_genre = games["genres"].str.contains(genre) == True
    return games[has_genre]["price"].median()

result = median_price(games, "Indie")
print("Median Indie price:", result)
```
